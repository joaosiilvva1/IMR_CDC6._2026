```python
import math
import time

print("--- INICIANDO LABORATÓRIO 2 (SIMULAÇÃO ROS) ---")


angular_speed = 0.5  
target_angle = math.pi / 2.0  

tempo_necessario = target_angle / angular_speed
print(f"[INFO] Velocidade angular: {angular_speed} rad/s")
print(f"[INFO] Ângulo alvo a ser atingido: {target_angle:.4f} radianos (90 graus)")
print(f"[INFO] Tempo estimado calculado para o giro: {tempo_necessario:.2f} segundos")

print("\n[INFO] Publicando velocidade no tópico /cmd_vel...")
t0 = time.time()
current_angle = 0.0

while current_angle < target_angle:
    time.sleep(0.1) # Simula o delay do ROS (10Hz)
    t1 = time.time()
    current_angle = angular_speed * (t1 - t0)
    print(f" -> Robô girando... Orientação atual: {current_angle:.2f} rad", end="\r")


print(f"\n -> Robô girando... Orientação atual: {target_angle:.2f} rad")
print("\n[INFO] Enviando comando de emergência / zeramento: vel_msg.angular.z = 0.0")
print("[INFO] Giro concluído com sucesso! Robô parou no próprio eixo mantendo a posição fixa.")
```
