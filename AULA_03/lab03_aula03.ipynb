```python
import pygame
import math
import numpy as np

LARGURA, ALTURA = 900, 650

class RobotLab3:
    def __init__(self, x, y):
        self.x, self.y, self.theta = float(x), float(y), 0.0
        self.sensor_angles = [-math.pi/2, -math.pi/4, 0.0, math.pi/4, math.pi/2]
        self.sensor_range = 200.0
        self.sensor_readings = [self.sensor_range] * 5

    def cast_rays(self, obstacles):
        self.sensor_readings = []
        for beta in self.sensor_angles:
            angle = self.theta + beta
            min_dist = self.sensor_range
            
            for step in range(5, int(self.sensor_range), 4):
                rx = self.x + step * math.cos(angle)
                ry = self.y + step * math.sin(angle)
                
                if rx <= 0 or rx >= LARGURA or ry <= 0 or ry >= ALTURA:
                    min_dist = float(step)
                    break
                
                if any(obs.collidepoint(rx, ry) for obs in obstacles):
                    min_dist = float(step)
                    break
            
            ruido = np.random.normal(0, 2.0)
            dist_ruidosa = max(0.0, min(self.sensor_range, min_dist + ruido))
            self.sensor_readings.append(dist_ruidosa)

    def draw(self, surface, font):
        for i, beta in enumerate(self.sensor_angles):
            angle = self.theta + beta
            dist = self.sensor_readings[i]
            rx = self.x + dist * math.cos(angle)
            ry = self.y + dist * math.sin(angle)
            
            pygame.draw.line(surface, (0, 255, 100), (int(self.x), int(self.y)), (int(rx), int(ry)), 2)
            text = font.render(f"{dist:.1f}", True, (255, 255, 255))
            surface.blit(text, (int(rx) + 5, int(ry) + 5))
            
        pygame.draw.circle(surface, (0, 200, 255), (int(self.x), int(self.y)), 16)

def main_lab3():
    pygame.init()
    screen = pygame.display.set_mode((LARGURA, ALTURA))
    font = pygame.font.SysFont("monospace", 12)
    robot = RobotLab3(450, 325)
    obstacles = [pygame.Rect(300, 150, 100, 300), pygame.Rect(600, 200, 150, 150)]
    
    running = True
    while running:
        for event in pygame.event.get():
            if event.type == pygame.QUIT: running = False
            
        mx, my = pygame.mouse.get_pos()
        robot.theta = math.atan2(my - robot.y, mx - robot.x)
        robot.x += (mx - robot.x) * 0.02
        robot.y += (my - robot.y) * 0.02
        
        robot.cast_rays(obstacles)
        screen.fill((20, 24, 30))
        for obs in obstacles: pygame.draw.rect(screen, (180, 50, 50), obs)
        robot.draw(screen, font)
        
        pygame.display.flip()
        pygame.time.Clock().tick(60)
    pygame.quit()

if __name__ == "__main__": main_lab3()
```
